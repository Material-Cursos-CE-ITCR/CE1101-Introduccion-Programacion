### ¿Qué es iteración
Repetir un bloque del código de acuerdo a una condición.   
Existen 3 maneras para realizar una iteración:
1. Recursividad [taller de recursividad](https://github.com/Material-Cursos-CE-ITCR/CE1101-Introduccion-Programacion/blob/main/Talleres/notebooks_PracticaListasRecursividad.ipynb)
2. Ciclo `While`
3. Ciclo `For`

#### Ciclo while
```python
while(condition):
    #Bloque de código que se desea repetir.
#Continua aquí después al no cumplirse la condición.
``` 
El bloque de código delimitado por la identación se ejecuta mientras una condición sea verdadera. Una forma de utilizarlo es la repetición infinita conocido como: ```while True```.
Ejemplo del ciclo ```while```:  
```python
i = 1
while (i < 6):
  print(i)
  i += 1
```

#### Ciclo for-in 
```python
for item in List: #List debe ser un elemento iterable.
    #Bloque de código que se desea repetir.
#Continua aquí al finalizar los elementos de la lista.
``` 
Se utiliza cuando queremos repetir un bloque de código una cantidad finita de veces o necesitamos acceder a los elementos de una estructura de datos.
Ejemplo:
```python
for i in range(6): # range(10) -> crea una lista con valores del 0 al 9
  print(i)
```

#### Declaración Break
Nos permite cancelar la ejecución del ciclo sin importar que la condición siga siendo verdadera.

#### Declaración continue (no tan usada)
Nos permite cancelar la ejecución del ciclo actual y continuar en el siguiente ciclo. Sin ejecutar el resto del código una vez llamda la función.
